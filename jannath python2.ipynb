## Q2

```python
meters={
   "M001":{"last_month":120,"this_month":280},
   "M002":{"last_month":95,"this_month":110},
   "M003":{"last_month":60,"this_month":150},
   "M004":{"last_month":200,"this_month":210},
}
```

```python
flagged=[]
```

```python
for meter_id, data in meters.items():
    last=data["last_month"]
    current=data["this_month"]
    if current > (last *2):
        print(f"{meter_id}: {data['last_month']}to {data['this_month']} - FLAGGED")
        flagged.append(meter_id)
    else:
         print(f"{meter_id}: {data['last_month']} to{data['this_month']} - NORMAL")
             
print("\nFlagged Meters:",",".join(flagged))
print("Total Flagged:",len(flagged))
```

### Q3

```python
downtime_logs=[
     {"machine":"M1","shift":"Morning","downtime_min":45,"reason":"Mechanical","hourly_output_value":2000},
     {"machine":"M2","shift":"Night","downtime_min":90,"reason":"Electrical","hourly_output_value":1800},
     {"machine":"M1","shift":"Evening","downtime_min":20,"reason":"Mechanical","hourly_output_value":2000},
     {"machine":"M3","shift":"Morning","downtime_min":120,"reason":"Material Shortage","hourly_output_value":1500},
     {"machine":"M2","shift":"Morning","downtime_min":30,"reason":"Elecrical","hourly_output_value":1800},
     {"machine":"M1","shift":"Night","downtime_min":60,"reason":"Mechanical","hourly_output_value":2000},]
     
```

```python
total_minutes=0
total_cost=0

machine_downtime={}
reason_cost={}
machine_cost={}
```

```python
for log in downtime_logs:
    minutes=log["downtime_min"]
    machine=log["machine"]
    reason=log["reason"]
    
    cost = (minutes / 60) * log["hourly_output_value"]

    total_minutes += minutes
    total_cost += cost

    machine_downtime[machine]=machine_downtime.get(machine,0)+minutes
    reason_cost[reason]=reason_cost.get(reason,0)+cost
    machine_cost[machine]=machine_cost.get(machine,0)+cost

highest_machine=max(machine_cost,key=machine_cost.get)
highest_reason=max(reason_cost,key=reason_cost.get)
```

```python
print("Total Downtime:",total_minutes,
"minutes")
print("Total Cost:Rs.",total_cost)
print("Machine Downtime:", machine_downtime)
print("Most Costly Reason:",highest_reason,"Rs.",reason_cost[highest_reason])
print("Machine With Highest Cost:",
highest_machine,"Rs.",machine_cost[highest_machine])
      
print("Recommendation:Prioritize mechanical maintenance for M1 to reduce downtime costs.")
```
